```python
import json

with open("./output/abstract_generation/LoRA-Composition/output-Composition-topk1-k3-epoch1-aggtemp1-greedy-recentNone.json", 'r') as f:
    data = json.load(f)['golds']
 
data_psd = []   
for s in data:
    # print(s['generation'])
    sl = s['generation'].split('.')
    sl_psd = list(set(sl))
    sl_psd.sort(key=sl.index)
    if sl_psd[-1]:
        sl_psd = sl_psd[:-1]
    s_psd = '.'.join(sl_psd)
    s['generation'] = s_psd
    data_psd.append(s)
    
final_data = {"task": "abstract_generation", "golds": data_psd}
with open("./output/abstract_generation/LoRA-Composition/output-Composition-topk1-k3-epoch1-aggtemp1-greedy-recentNone-filtered.json", 'w') as f:
    json.dump(final_data, f, indent=4)
    
```

```python
import io
import contextlib

code_str = """

def generate_headline(article):
    return "Dallas Morning News, Anti-Abortion Groups Distance Themselves From Kristen Hatten"

article = "The Dallas Morning News and anti-abortion groups are distancing themselves from Kristen Hatten."
print("Your generated headline here")
print(generate_headline(article))

"""

# 创建一个 StringIO 来捕获输出
output_buffer = io.StringIO()

# 重定向 stdout
with contextlib.redirect_stdout(output_buffer):
    exec(code_str)

# 获取输出内容
output = output_buffer.getvalue()
if '.' in output:
    output_ls = output.split('.')
    ol_psd = list(set(output_ls))
    ol_psd.sort(key=output_ls.index)
    if ol_psd[-1] and len(output_ls) != len(ol_psd):
        ol_psd = ol_psd[:-1]
    output = '.'.join(ol_psd)

# print("捕获到的内容：")
print(output)
```
