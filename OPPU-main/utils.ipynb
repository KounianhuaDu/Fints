```python
import os
import re

base_dir = "/inspire/hdd/project/multi-agent/zhangweinan-24046/PA/OPPU-main/ckpt/pwab"

for name in os.listdir(base_dir):
    old_path = os.path.join(base_dir, name)
    if not os.path.isdir(old_path):
        continue  # 只处理文件夹
    
    # 去掉结尾的 -数字
    new_name = re.sub(r"-\d+$", "", name)
    new_path = os.path.join(base_dir, new_name)

    if new_name != name:
        if os.path.exists(new_path):
            print(f"⚠️ 目标已存在，跳过: {new_path}")
        else:
            os.rename(old_path, new_path)
            print(f"✅ 已重命名: {old_path} -> {new_path}")
```

```python
import os
import shutil

base_dir = "ckpt/abstract_generation"  # 可以改为你的目标目录路径

# 1. 获取所有 a-b 格式的文件夹
folders = [f for f in os.listdir(base_dir) if os.path.isdir(os.path.join(base_dir, f)) and "-" in f]

groups = {}
for folder in folders:
    try:
        a, b = folder.rsplit("-", 1)
        b = int(b)
    except ValueError:
        continue  # 跳过不符合格式的
    if a not in groups:
        groups[a] = []
    groups[a].append((b, folder))

# 2. 对每组只保留 b 最大的文件夹
for a, items in groups.items():
    items.sort(reverse=True)  # 按 b 降序
    max_b, max_folder = items[0]

    # 删除/忽略其他较小的文件夹
    for b, folder in items[1:]:
        path = os.path.join(base_dir, folder)
        print(f"Deleting {path}")
        shutil.rmtree(path)

    # 3. 重命名最大值文件夹为 a
    src = os.path.join(base_dir, max_folder)
    dst = os.path.join(base_dir, a)
    if src != dst:  # 避免重复重命名
        if os.path.exists(dst):
        # if False:
            shutil.rmtree(dst)
        print(f"Renaming {src} -> {dst}")
        os.rename(src, dst)
```
